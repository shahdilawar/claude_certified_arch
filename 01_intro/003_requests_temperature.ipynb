```python
# LOAD ENV VARIABLES
from dotenv import load_dotenv
import os
load_dotenv()
api_key = os.getenv("ANTHROPIC_API_KEY")
```

```python
from anthropic import Anthropic
client = Anthropic(api_key=api_key)
model = "claude-haiku-4-5-20251001"
```

```python
#Initialize messages list
messages = []
# Add the user messages to the list
def add_user_message(messages, text):
    messages.append({
        "role": "user",
        "content": text
    })

# Add the assistant messages to the list
def add_assistant_message(messages, text):
    messages.append({
        "role": "assistant",
        "content": text
    })

# Create a conversation loop
parameters = {
    "model": model,
    "max_tokens": 1000,
    "messages": messages
}

def chat(messages, system=None, temperature=0.7):
    if system:
        parameters["system"] = system
    
    parameters["temperature"] = temperature

    messages = client.messages.create(**parameters)
    return messages.content[0].text
```

```python

add_user_message(messages, "Provide me a movie idea")

answer = chat(messages, system=None, temperature = 0.0)
print(answer)
```
