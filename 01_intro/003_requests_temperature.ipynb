```python
# LOAD ENV VARIABLES
from dotenv import load_dotenv
import os
load_dotenv()
api_key = os.getenv("ANTHROPIC_API_KEY")
```

```python
from anthropic import Anthropic
client = Anthropic(api_key=api_key)
model = "claude-haiku-4-5-20251001"
```

```python
#Initialize messages list
messages = []
# Helper method to add the user messages to the list
def add_user_message(messages, text):
    messages.append({
        "role": "user",
        "content": text
    })

# Helper method to add the assistant messages to the list
def add_assistant_message(messages, text):
    messages.append({
        "role": "assistant",
        "content": text
    })

# Create a parameters dictionary for the API request. This way System message can be added to the parameters dictionary if needed.
parameters = {
    "model": model,
    "max_tokens": 1000,
    "messages": messages
}

# Create a conversation loop
def chat(messages, system=None, temperature=0.7):
    if system:
        parameters["system"] = system
    
    parameters["temperature"] = temperature

    messages = client.messages.create(**parameters)

    # Extract the text from the response
    for block in messages.content:
        if block.type == "text":
            return block.text
    return ""
```

```python

add_user_message(messages, "Provide me a movie idea")

response = chat(messages, system = None, temperature = 1.0)
print(response)
```
