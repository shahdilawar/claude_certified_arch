```python
# LOAD ENV VARIABLES
from dotenv import load_dotenv
import os
load_dotenv()
api_key = os.getenv("ANTHROPIC_API_KEY")
```

```python
from anthropic import Anthropic
client = Anthropic(api_key=api_key)
model = "claude-sonnet-5"
```

```python
# Helper method to add the user messages to the list
def add_user_message(messages, text):
    messages.append({
        "role": "user",
        "content": text
    })

# Helper method to add the assistant messages to the list
def add_assistant_message(messages, text):
    messages.append({
        "role": "assistant",
        "content": text
    })

# Create a conversation loop
def chat(messages):
    messages = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )
    for block in messages.content:
        if block.type == "text":
            return block.text
    return ""
```

```python
# Getting user input in a while loop
messages = []
while True:
    user_input = input("You: ")
    add_user_message(messages, user_input)
    response = chat(messages)
    add_assistant_message(messages, response)  
    # print the generated response from the model
    print("------------")
    print(response)
    print("------------")
    
```
