```python
# LOAD ENV VARIABLES
from dotenv import load_dotenv
import os
load_dotenv()
api_key = os.getenv("ANTHROPIC_API_KEY")
```

```python
from anthropic import Anthropic
client = Anthropic(api_key=api_key)
model = "claude-haiku-4-5-20251001"     
```

```python
# Add the user messages to the list
def add_user_message(messages, text):
    messages.append({
        "role": "user",
        "content": text
    })

# Add the assistant messages to the list
def add_assistant_message(messages, text):
    messages.append({
        "role": "assistant",
        "content": text
    })

# Create a conversation loop
def chat(messages, stop_sequences=None):
    messages = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )
    return messages.content[0].text
```

```python
messages = []

prompt = """
Generate three different sample AWS CLI commands. Each should be very short.
"""

add_user_message(messages, prompt)
add_assistant_message(messages, "Here are all three commands without comments:\n'''bash")

answer = chat(messages, stop_sequences=["'''"])
print(answer.strip())

```
