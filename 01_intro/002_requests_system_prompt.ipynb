```python
# LOAD ENV VARIABLES
from dotenv import load_dotenv
import os
load_dotenv()
api_key = os.getenv("ANTHROPIC_API_KEY")
```

```python
from anthropic import Anthropic
client = Anthropic(api_key=api_key)
model = "claude-sonnet-5"
```

```python
#Initialize messages list
messages = []
# Helper method to add the user messages to the list
def add_user_message(messages, text):
    messages.append({
        "role": "user",
        "content": text
    })

# Helper method to add the assistant messages to the list
def add_assistant_message(messages, text):
    messages.append({
        "role": "assistant",
        "content": text
    })

# Set the parameters dictionary for the API request. This way System message can be added to the parameters dictionary if needed.
parameters = {
    "model": model,
    "max_tokens": 1000,
    "messages": messages
}

# Create a conversation loop
def chat(messages, system = None):
    if system:
        parameters["system"] = system
    messages = client.messages.create(**parameters)

    for block in messages.content:
        if block.type == "text":
            return block.text
    return ""
```

```python
add_user_message(messages, "How do I solve 5x + 2 = 3 for x?")

system = '''
You are a patient math tutor. Do not directly give the answer. 
Guide them to a solution step by step
'''

answer = chat(messages, system)
print(answer)
```
