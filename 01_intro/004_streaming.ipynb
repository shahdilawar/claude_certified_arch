```python
# LOAD ENV VARIABLES
from dotenv import load_dotenv
import os
load_dotenv()
api_key = os.getenv("ANTHROPIC_API_KEY")
```

```python
from anthropic import Anthropic
client = Anthropic(api_key=api_key)
model = "claude-haiku-4-5-20251001"
```

```python
#initialize empty messages list
messages = []
#add user message.
def add_user_message(messages, content):
    messages.append({"role": "user", "content": content})
```

```python
user_msg = "Write a 5 sentence description of a fake database"

add_user_message(messages, user_msg)

stream = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=messages,
    stream=True
)

for event in stream:
    print(event)
```

```python
user_msg = "Write a 5 sentence description of a fake database"

add_user_message(messages, user_msg)

with client.messages.stream(
    model=model,
    max_tokens=1000,
    messages=messages
) as stream:
    for text in stream.text_stream:
        print(text)
```

```python
#Getting the full response after streaming to save it in a variable.    
user_msg = "Write a 5 sentence description of a fake database"  

add_user_message(messages, user_msg)

with client.messages.stream(    
    model=model,
    max_tokens=1000,
    messages=messages
) as stream:
    full_response = ""
    for text in stream.text_stream:
        pass

    # Get complete response after streaming is done
    full_response = stream.get_final_message()
    print(full_response)
```
