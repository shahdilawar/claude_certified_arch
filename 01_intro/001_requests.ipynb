```python
# LOAD ENV VARIABLES
from dotenv import load_dotenv
import os
load_dotenv()
api_key = os.getenv("ANTHROPIC_API_KEY")
```

```python
from anthropic import Anthropic
client = Anthropic(api_key=api_key)
model = "claude-haiku-4-5-20251001"
```

```python
messages = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence"
          }
    ]
)
```

```python
messages.content[0].text # print the response from the model
```

```python
# Add the user messages to the list
def add_user_message(messages, text):
    messages.append({
        "role": "user",
        "content": text
    })

# Add the assistant messages to the list
def add_assistant_message(messages, text):
    messages.append({
        "role": "assistant",
        "content": text
    })

# Create a conversation loop
def chat(messages):
    messages = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )
    return messages.content[0].text
```

```python
# Start with an empty list
conversation = []

# Add the first user message
add_user_message(conversation, 
    "What is quantum computing? Answer in one sentence")

# Get the assistant's response
response = chat(conversation)
print(f"message is :  {response}")

# Add claude's response to the conversation
add_assistant_message(conversation, response)

# Add another user message
add_user_message(conversation, "Write another sentence.")

# Get the assistant's response
response = chat(conversation)
print(f"followup message is :  {response}")
```
